```python
# Import required libraries

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

```

```python
# Load processed classification data

X_train = pd.read_csv('../data/X_train_classification.csv')
X_test = pd.read_csv('../data/X_test_classification.csv')

y_train = pd.read_csv('../data/y_train_classification.csv').squeeze()
y_test = pd.read_csv('../data/y_test_classification.csv').squeeze()

print("Classification data loaded successfully.")
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
# Verify classification data

print("Training shape:", X_train.shape)
print("Testing shape :", X_test.shape)

print("\nMissing values in training data:", X_train.isnull().sum().sum())
print("Missing values in testing data :", X_test.isnull().sum().sum())

print("\nTarget distribution:")
print(y_train.value_counts())
```

```python
# Import classification models and evaluation metrics

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score
)

print("Classification models and metrics imported.")
```

```python
# Train Logistic Regression model

logistic_model = LogisticRegression(
    max_iter=1000,
    class_weight='balanced',
    random_state=42
)

logistic_model.fit(X_train, y_train)

print("Logistic Regression trained successfully.")
```

```python
# Evaluate Logistic Regression

logistic_pred = logistic_model.predict(X_test)

logistic_accuracy = accuracy_score(y_test, logistic_pred)
logistic_f1 = f1_score(y_test, logistic_pred, average='weighted')

print("Logistic Regression Results")
print("Accuracy:", logistic_accuracy)
print("Weighted F1:", logistic_f1)

print("\nClassification Report:")
print(classification_report(y_test, logistic_pred))
```

```python
# Train Random Forest classifier

rf_classifier = RandomForestClassifier(
    n_estimators=100,
    max_depth=15,
    class_weight='balanced',
    random_state=42,
    n_jobs=-1
)

rf_classifier.fit(X_train, y_train)

print("Random Forest classifier trained successfully.")
```

```python
# Evaluate Random Forest classifier

rf_pred = rf_classifier.predict(X_test)

rf_accuracy = accuracy_score(y_test, rf_pred)
rf_f1 = f1_score(y_test, rf_pred, average='weighted')

print("Random Forest Results")
print("Accuracy:", rf_accuracy)
print("Weighted F1:", rf_f1)

print("\nClassification Report:")
print(classification_report(y_test, rf_pred))
```

```python
# Compare classification models

classification_results = pd.DataFrame({
    'Model': ['Logistic Regression', 'Random Forest'],
    'Accuracy': [logistic_accuracy, rf_accuracy],
    'Weighted F1': [logistic_f1, rf_f1]
})

print(classification_results)
```
